```python
# Calcular conteos
conteos = df.groupby(['¿Probó fundas Titan durante la visita?', 'Resultado de prueba en obra']).size().reset_index(name='frecuencias')

# Calcular el total para obtener proporciones
total = conteos['frecuencias'].sum()

# Crear columna con proporción decimal (frecuencia / total)
conteos['proporcion'] = (conteos['frecuencias'] / total).round(2)  # redondeado a 2 decimales

# Ordenar por proporción descendente
resultado = conteos.sort_values(by='proporcion', ascending=False)

resultado
```
