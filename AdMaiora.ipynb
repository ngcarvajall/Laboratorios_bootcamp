```python
# Importar librerías
from docx import Document
from docx.shared import Inches

# Crear función para generar planificación semanal en formato de tabla
def generar_planificacion_semanal():
    # Crear documento nuevo
    doc = Document()
    doc.add_heading('Planificación Semanal', level=1)
    
    # Encabezado general
    semana = "Semana: 3 - 31 de Marzo"
    curso = "Curso: 1ro de Primaria"
    materia = "Materia: Lengua Española"
    profesor = "Profesor/a: Daniela Cabrera"
    doc.add_paragraph(f"{semana}        {curso}        {materia}        {profesor}")
    
    # Competencias fundamentales
    competencias_fundamentales = (
        "Competencias fundamentales: Expresa y comprende, de forma oral y escrita en diferentes contextos, "
        "textos funcionales y literarios muy sencillos y de estructura sintáctica simple, mediante el uso de medios y "
        "recursos apropiados, demostrando avance progresivo en sus procesos de lectura y escritura con cierto nivel de autonomía."
    )
    doc.add_paragraph(competencias_fundamentales)
    
    # Datos para cada día de la semana
    dias_semana = [
        {
            "fecha": "Lunes, 3 de Marzo",
            "competencias_especificas": "Expresa y comprende textos funcionales y literarios sencillos.",
            "contenido": "El cuento: Componentes y estructura (inicio, nudo y desenlace).",
            "indicadores_logros": "Comprende y produce textos literarios sencillos.",
            "secuencia_actividades": (
                "Inicio: Retroalimentación de contenido.\n"
                "Desarrollo: La maestra preguntará: ¿qué es un títere? ¿Para qué se utiliza?\n"
                "Cierre: Retroalimentación de contenido."
            ),
            "ejes_tematicos": "Matemáticas, Naturales",
            "tipo_evaluacion": "Oral",
            "asignaciones": "Crea tu propio títere y realiza un video contando tu historia."
        },
        {
            "fecha": "Martes, 4 de Marzo",
            "competencias_especificas": "Expresa y comprende textos funcionales y literarios sencillos.",
            "contenido": "El cuento: Componentes y estructura (inicio, nudo y desenlace).",
            "indicadores_logros": "Comprende y produce textos literarios sencillos.",
            "secuencia_actividades": (
                "Inicio: Retroalimentación de contenido.\n"
                "Desarrollo: En grupos, crearán un cuento con inicio, nudo y desenlace.\n"
                "Cierre: Revisión de presentaciones."
            ),
            "ejes_tematicos": "Ciencias Sociales",
            "tipo_evaluacion": "Oral",
            "asignaciones": "Dibujar una escena del cuento creado."
        }
    ]
    
    # Crear una tabla con encabezados
    tabla = doc.add_table(rows=1, cols=8)
    tabla.style = 'Table Grid'
    encabezados = ["Fecha", "Competencias Específicas", "Contenidos", "Indicadores de Logros", 
                   "Secuencia de Actividades", "Ejes Temáticos Transversales", "Tipo de Evaluación", "Asignaciones"]
    hdr_cells = tabla.rows[0].cells
    for i, encabezado in enumerate(encabezados):
        hdr_cells[i].text = encabezado
    
    # Agregar filas con la información de cada día
    for dia in dias_semana:
        fila = tabla.add_row().cells
        fila[0].text = dia["fecha"]
        fila[1].text = dia["competencias_especificas"]
        fila[2].text = dia["contenido"]
        fila[3].text = dia["indicadores_logros"]
        fila[4].text = dia["secuencia_actividades"]
        fila[5].text = dia["ejes_tematicos"]
        fila[6].text = dia["tipo_evaluacion"]
        fila[7].text = dia["asignaciones"]
    
    # Guardar el documento
    doc.save('planificacion_semanal_tabla.docx')
    print("¡Planificación semanal en formato de tabla generada con éxito!")

# Ejecutar la función
generar_planificacion_semanal()
```
